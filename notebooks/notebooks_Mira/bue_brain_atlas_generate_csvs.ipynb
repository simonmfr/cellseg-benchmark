```python
import numpy as np
import pandas as pd
```

```python
# adjust filepath to required input csv: /dss/dssfs03/pn52re/pn52re-dss-0001/cellseg-benchmark/misc/celltype_composition/bluebrainatlas

bs_interbrain_counts = pd.read_csv("BS_Interbrain_counts.csv", sep=";")
cerebrum_counts = pd.read_csv("Cerebrum_all_cell_counts.csv", sep=";")
ft_lateral_medial_counts = pd.read_csv("FT_lateral_medial_counts.csv", sep=";")

bs_interbrain_density = pd.read_csv("BS_Interbrain_density.csv", sep=";")
cerbrum_density = pd.read_csv("Cerebrum_density.csv", sep=";")
ft_lateral_medial_density = pd.read_csv("FT_lateral_medial_density.csv", sep=";")
```

```python
areas_counts = {"brain_stem": bs_interbrain_counts, "cerebrum": cerebrum_counts, "fibre_tracts": ft_lateral_medial_counts}

rows = []

for area,counts in areas_counts.items():
    row = counts.iloc[0].to_dict()
    row["Area"] = area
    rows.append(row)

counts = pd.DataFrame(rows)
counts

col = counts.pop("Area")
counts.insert(0, "Area", col)
counts.fillna(0, inplace=True)
counts = counts.astype({col: int for col in counts.columns if col != "Area"})

counts.to_csv("blue_brain_atlas_counts.csv")
counts
```

```python
areas_density = {"brain_stem": bs_interbrain_density, "cerebrum": cerbrum_density, "fibre_tracts": ft_lateral_medial_density}

densities = []

for area,df in areas_density.items():
    row = df.iloc[0].to_dict()
    row["Area"] = area
    densities.append(row)

area_densities = pd.DataFrame(densities)

col = area_densities.pop("Area")
area_densities.insert(0, "Area", col)
area_densities.fillna(0, inplace=True)

area_densities.to_csv("blue_brain_atlas_density.csv")
area_densities
```
